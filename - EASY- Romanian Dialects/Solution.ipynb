```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

SUBTASK 1

```python
result_subtask1 = 0
for i in range(0,len(train_df)):
    words = [word.strip(".,!?\"';:") for word in train_df['text'].iloc[i].split()]
    for x in words:
        if x == 'pâni':
            result_subtask1 += 1

for i in range(0,len(test_df)):
    words = [word.strip(".,!?\"';:") for word in test_df['text'].iloc[i].split()]
    for x in words:
        if x == 'pâni':
            result_subtask1 += 1
```

SUBTASK 2

SUBTASK3

```python
subtask3_answer = []
for i in range(0,len(test_df)):
    words = [word.strip(".,!?\"';:").lower() for word in test_df['text'].iloc[i].split()]
    re = 0
    for word in words:
        for s in range(0,len(word)):
            if word[s] in ['ă', 'â', 'î', 'ș', 'ț']: re += 1
    subtask3_answer.append(re)
```

SUBTASK 4

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC

vectorizer = TfidfVectorizer(analyzer = 'char',ngram_range = (1,7))
model = LinearSVC(C = 1.001,random_state = 42, max_iter = 15000)

model.fit(vectorizer.fit_transform(train_df['text'].values), train_df['label'])

subtask4_answers = model.predict(vectorizer.transform(test_df['text'].values))
```

SUBMISSION

```python
df1 = pd.DataFrame({
    'subtaskID' : 1,
    'datapointID' : 1,
    'answer' : [result_subtask1]
})

df2 = pd.DataFrame({
    'subtaskID' : 2,
    'datapointID' : 1,
    'answer' : [result_subtask1]
})

df3 = pd.DataFrame({
    'subtaskID' : 3,
    'datapointID' : test_df['ID'],
    'answer' : subtask3_answer
})

df4 = pd.DataFrame({
    'subtaskID' : 4,
    'datapointID' : test_df['ID'],
    'answer' : subtask4_answers
})

pd.concat([df1,df2,df3,df4]).to_csv('submission.csv',index = False)
```
