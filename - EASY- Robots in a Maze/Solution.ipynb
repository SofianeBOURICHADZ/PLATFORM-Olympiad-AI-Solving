```python
import numpy as np
import pandas as pd

train_df = pd.read_csv('train.csv')
test_df = pd.read_csv('test.csv')
```

SUBTASK 5

```python
from sklearn.ensemble import HistGradientBoostingClassifier

train_y = train_df['strategy_label']
train_x = train_df.drop(['strategy_label','robot_id'],axis = 1)
train_x['arena_type'] = train_x['arena_type'].map({'urban':0,'forest':1,'desert':2,'ice':3})
train_x['weather'] = train_x['weather'].map({'clear':0,'fog':1,'windy':2})

model = HistGradientBoostingClassifier(learning_rate = 0.02,max_iter = 3000,max_depth = 7,random_state = 42,l2_regularization = 1.0,)
model.fit(train_x,train_y)
test_x = test_df.drop(['robot_id'],axis = 1)
test_x['arena_type'] = test_x['arena_type'].map({'urban':0,'forest':1,'desert':2,'ice':3})
test_x['weather'] = test_x['weather'].map({'clear':0,'fog':1,'windy':2})

subtask5_results = model.predict(test_x)
```

SUBMISSION

```python
p1 = pd.DataFrame({
    'robot_id':['GLOBAL'] * 4,
    'subtaskID': [1,2,3,4],
    'answer': [len(train_df['arena_type'].unique()), np.max(train_df['avg_speed_mps'].values), list(dict(train_df['arena_type'].value_counts()).keys())[0], np.max(train_df['items_collected'].values)]
})

p2 = pd.DataFrame({
    'robot_id': test_df['robot_id'],
    'subtaskID': 5,
    'answer': subtask5_results
})

pd.concat([p1,p2]).to_csv('submission.csv')
```
